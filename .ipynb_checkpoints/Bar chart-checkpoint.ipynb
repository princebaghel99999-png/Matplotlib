```python
import matplotlib.pyplot as plt
```

```python
students = ['Ram','Shayam','Mohan']
marks = [80,90,70]

plt.bar(students,marks)

plt.show()
```

```python
#Title aur labels

plt.bar (students,marks)
plt.title('Student marks')
plt.xlabel('Student')
plt.ylabel('Marks')

plt.show()
```

```python
# bar color
plt.bar(students,marks, color='red')
plt.show()
```

```python
# bar width
plt.bar(
    students,
    marks,
    width=0.5
)

plt.show()
```

```python
# grid add karna

plt.bar(students,marks)

plt.grid(axis='y' ) # bar chars me mostly axis='y' use hota hai

plt.show()
```

```python
students= ['Ram','Shayam','Mohan']

marks= [80,90,70]

bars = plt.bar(students,marks)

plt.bar_label(bars,padding=3) # value automatically bar ke upar
plt.show()
```

```python
# horizontal bar chart

plt.barh(students,marks) # horizontal bar chart

plt.show()
```

```python
# multiple colors 

colors= ['red','green','blue']

plt.bar(
    students,
    marks,
    color=colors
)

plt.show()
```

```python
# figure size 
plt.figure(figsize=(8,5))

plt.bar(students,marks)

plt.show()
```

```python

# Grouped Bar Chart
# Grouped Bar Chart ka use ek hi category ke andar multiple data sets ko compare karne ke 
#liye kiya jata hai, 
# jaise Students ke Maths aur Science marks.


import numpy as np

students = ['Ram','Shayam','Mohan']

maths = [80,90,70]
science = [75,85,65]

x= np.arange(len(students))

plt.bar(x-0.2,maths, width=0.4, label='Maths')
plt.bar(x+0.2, science, width=0.4, label='Science')

plt.xticks(x,students)

plt.legend()

plt.show()

# Ek student ke liye 2 subjects compare kar sakte ho.
# x = [0,1,2] → categories ki positions
# x-0.2 → left shift
# x+0.2 → right shift
# width=0.4 → bar ki width
# plt.xticks() → numbers ko names me badalta hai
# plt.legend() → Maths aur Science ka label dikhata hai
```

```python
# stacked bar chart

students = ['Ram','Shayam','Mohan']

maths = [80,90,70]
science = [20,10,15]

plt.bar(students,maths,label='Maths')

plt.bar(
    students,
    science,
    bottom= maths,
    label = 'Science'
)

plt.legend()
plt.show()

# yahan bars ek doosre ke upar stack hoti hain
```

```python
# pandas ke sath

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Student':['Ram','Shayam','Mohan'],
    'Marks':[80,90,70]
})

df.plot(
    kind='bar',
    x='Student',
    y='Marks'
)

plt.show()
```

```python
months =['Jan','Feb','Mar','Apr','May']
sales = [10000,15000,18000,25000,35000]

bars = plt.bar(months,sales)

plt.bar_label(bars)

plt.title("Monthly Sales Report")
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
```
