```python
# # Pie Chart Kya Hota Hai?

# # Pie Chart ek circular chart hota hai jo data ko percentage ya parts me dikhata hai.

# # Iska use tab karte hain jab hume dekhna ho:

# # Total me kis category ka kitna contribution hai.
# Example

# Ek class me:

# Python = 40 students
# Java = 30 students
# MERN = 20 students
# Data Analytics = 10 students

# Total:

# 40 + 30 + 20 + 10 = 100

# Pie Chart dikhayega:

# Python = 40%
# Java = 30%
# MERN = 20%
# Data Analytics = 10%
```

```python
import matplotlib.pyplot as plt
```

```python
courses = ['Python',"Java",'MERN','Data Analytics']

students= [40,30,20,10]

plt.pie(students)

plt.show()
```

```python
# label add karna
# agar labels nahi denge to sirf slices dikhenge har slice ke sath naam dikhega

plt.pie(
    students,
    labels=courses
)
plt.show()
```

```python
# percentage show karna
```

```python
plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%'
)

plt.show()
```

```python
courses = ['Python',"Java",'MERN','Data Analytics']

students= [40,30,20,10]

plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%'
)

plt.title('Course Enrollment Distribution')

plt.show()
```

```python
# explode kisi slice ko alag karna ho 

explode = [0,0.1,0,0]

plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    explode=explode
)
plt.show()
```

```python
# colors add karna

colors=['yellow','blue','green','orange']

plt.pie(
    students,
    labels=courses,
    colors=colors,
    autopct='%1.1f%%'
)
plt.show()
```

```python
# shadow add karna

plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    shadow=True
)
plt.show()
```

```python
# start angle

plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    startangle=90
)
plt.show()
```

```python
import matplotlib.pyplot as plt

labels=['Python','Java','Mern','Data Analytics']

values=[40,30,20,10]

plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%'
)
plt.title('Pie chart')
plt.show()
```
