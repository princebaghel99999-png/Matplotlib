```python
import numpy as np

students = ['Ram','Shayam','Mohan']

maths = [80,90,70]
science = [75,85,65]

x = np.arange(len(students))

plt.bar(x-0.2,maths,width=0.4, label='Maths')
plt.bar(x+0.2, science, width=0.4, label='Science')

plt.xticks(x,students)
plt.legend()
plt.show()
```

```python
# pie chart 
```

```python
import matplotlib.pyplot as plt
```

```python
courses = ['Python','Java','Mern','Data Analytics']

students = [40,30,20,10]

plt.pie(students)
plt.show()
```

```python
plt.pie(
    students,
    labels=courses
)
plt.show()
```

```python
plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%'
)
plt.title('Course Enrollment Distibution')
plt.show()
```

```python
explode = [0,0.1,0,0]

plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    explode = explode
)
plt.show()
```

```python
colors = ['yellow','blue','green','orange']
plt.pie(
    students,
    labels=courses,
    colors=colors,
    autopct='%1.1f%%'
)
plt.show()
```

```python
plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    shadow=True
)
plt.show()
```

```python
plt.pie(
    students,
    labels=courses,
    autopct='%1.1f%%',
    startangle=60
)
plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.subplot(1,2,1)
plt.plot([1,2,3],[4,5,6])

plt.subplot(1,2,2)
plt.plot([1,2,3],[6,5,4])

plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot([1,2,3],[1,2,3])

plt.subplot(2,2,2)
plt.plot([1,2,3],[3,2,1])

plt.subplot(2,2,3)
plt.plot([1,2,3],[2,3,1])

plt.subplot(2,2,4)
plt.plot([1,2,3],[1,3,2])

plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot([1,2,3],[1,2,3])
plt.title('Sales')
plt.xlabel('Months')
plt.ylabel('Amount')

plt.subplot(2,2,2)
plt.plot([1,2,3],[3,2,1])
plt.title('Profit')
plt.xlabel('Month')
plt.ylabel('Amount')

plt.subplot(2,2,3)
plt.plot([1,2,3],[2,3,1])
plt.title('Expenses')
plt.xlabel('Month')
plt.ylabel('Amount')

plt.subplot(2,2,4)
plt.plot([1,2,3],[1,3,2])
plt.title('Customers')
plt.xlabel('Month')
plt.ylabel('Count')

plt.tight_layout()

plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr']

sales = [100,150,200,250]
profit = [20,40,60,80]

plt.subplot(1,2,1)
plt.plot(months,sales)
plt.title('Sales')

plt.subplot(1,2,2)
plt.plot(months,profit)
plt.title('Profit')

plt.show()
```

```python
fig, ax = plt.subplots(2,2)

ax[0,0].plot([1,2,3],[1,2,3])
ax[0,1].plot([1,2,3],[3,2,1])

ax[1,0].plot([1,2,3],[2,3,1])
ax[1,1].plot([1,2,3],[1,3,2])

plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr']

sales = [100,150,200,250]
profit = [20,40,60,80]

fig, ax = plt.subplots(1,2, figsize=(10,5))

ax[0].plot(months,sales)
ax[0].set_title('Sales')

ax[1].plot(months,profit)
ax[1].set_title('Profit')

plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(10,8))

ax[0,0].plot([1,2,3,4],[10,20,15,25])
ax[0,0].set_title('Line Chart')

ax[0,1].bar(
    ['Jan','Feb','Mar','Apr'],
    [100,150,120,180]
)
ax[0,1].set_title('Bar Chart')

ax[1,0].pie(
    [40,30,20,10],
    labels=['Python','Java','MERN','DA'],
    autopct='%1.1f%%'
)
ax[1,0].set_title('Pie Chart')

ax[1,1].scatter(
    [1,2,3,4,5],
    [20,40,35,60,80]
)
ax[1,1].set_title('Scatter Plot')

plt.tight_layout()
plt.show()
```
